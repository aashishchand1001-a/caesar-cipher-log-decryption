### Dataset
Advertising spend $x$ (£000) against sales $y$ (£000) for six consecutive months:

| Month | Advertising Spend $x$ (£000) | Sales $y$ (£000) |
| :---: | :---: | :---: |
| 1 | 1 | 3 |
| 2 | 2 | 5 |
| 3 | 3 | 4 |
| 4 | 4 | 8 |
| 5 | 5 | 10 |
| 6 | 6 | 12 |

**Given:**
- Mean of $x$: $\bar{x} = 3.5$
- Mean of $y$: $\bar{y} = 7.0$


```python
# Define dataset using standard Python lists (No external libraries)
x = [1, 2, 3, 4, 5, 6]
y = [3, 5, 4, 8, 10, 12]
n = len(x)

# Calculate sample means
x_bar = sum(x) / n
y_bar = sum(y) / n

print(f"Sample size (n): {n}")
print(f"Calculated x̄: {x_bar:.4f} (Given: 3.5)")
print(f"Calculated ȳ: {y_bar:.4f} (Given: 7.0)")
```

## Part A — Correlation, the Line, Residuals & Interpretation

### Question 1: Calculate the Pearson Correlation Coefficient $r$

**Formulas:**
$$S_{xy} = \sum (x - \bar{x})(y - \bar{y})$$
$$S_{xx} = \sum (x - \bar{x})^2$$
$$S_{yy} = \sum (y - \bar{y})^2$$
$$r = \frac{S_{xy}}{\sqrt{S_{xx} \times S_{yy}}}$$


```python
# Question 1: Pearson correlation coefficient r

# Calculate sum of products and sums of squared deviations
Sxy = sum((xi - x_bar) * (yi - y_bar) for xi, yi in zip(x, y))
Sxx = sum((xi - x_bar) ** 2 for xi in x)
Syy = sum((yi - y_bar) ** 2 for yi in y)

# Pearson correlation coefficient r (using pure Python exponentiation for square root)
r = Sxy / ((Sxx * Syy) ** 0.5)

print("--- Question 1 Calculations ---")
print(f"Sxy = Σ(x - x̄)(y - ȳ) = {Sxy:.4f}")
print(f"Sxx = Σ(x - x̄)²       = {Sxx:.4f}")
print(f"Syy = Σ(y - ȳ)²       = {Syy:.4f}")
print(f"r   = Sxy / √(Sxx*Syy) = {r:.6f}")
print(f"Rounded r             = {round(r, 4)}")
```

### Question 2: Find the Least-Squares Regression Line

**Formulas:**
$$\text{Slope } b = \frac{S_{xy}}{S_{xx}}$$
$$\text{Intercept } a = \bar{y} - b\bar{x}$$
$$\text{Regression Equation: } \hat{y} = a + bx$$


```python
# Question 2: Slope and Intercept of Regression Line

# Slope b
b = Sxy / Sxx

# Intercept a
a = y_bar - (b * x_bar)

print("--- Question 2 Calculations ---")
print(f"Slope (b)     = Sxy / Sxx = {Sxy:.1f} / {Sxx:.1f} = {b:.6f} ≈ {b:.3f}")
print(f"Intercept (a) = ȳ - b(x̄)  = {y_bar:.1f} - ({b:.4f} * {x_bar:.1f}) = {a:.6f} ≈ {a:.3f}")
print(f"Regression Line Equation: ŷ = {a:.4f} + {b:.4f}x")
print(f"In rounded form:          ŷ = {round(a, 3)} + {round(b, 3)}x")
```

### Question 3: Calculate $R^2$ and Interpretation

**Formula:**
$$R^2 = r^2$$


```python
# Question 3: Coefficient of Determination R²
r_squared = r ** 2
r_squared_pct = r_squared * 100

print("--- Question 3 Calculations ---")
print(f"R² = r² = ({r:.6f})² = {r_squared:.6f}")
print(f"R² as percentage     = {r_squared_pct:.2f}% (≈ {round(r_squared_pct, 1)}%)")
```

**What does this percentage tell you about the variation in sales?**
- **Interpretation:** $R^2 = 91.43\%$ indicates that approximately **$91.4\%$ of the total variation in monthly sales ($y$) is explained by advertising expenditure ($x$)** via the linear regression model.
- Only the remaining **$8.57\%$ of the variation** is unexplained by the model (attributed to residual variance, random factors, or unobserved market conditions).


### Question 4: Calculate Adjusted $R^2$

**Formula:**
$$R^2_{\text{adj}} = 1 - (1 - R^2)\frac{n - 1}{n - k - 1}$$
where $n = 6$ (number of observations) and $k = 1$ (number of independent predictor variables).


```python
# Question 4: Adjusted R²
k = 1  # single predictor: advertising spend

r_squared_adj = 1 - (1 - r_squared) * (n - 1) / (n - k - 1)
r_squared_adj_pct = r_squared_adj * 100

print("--- Question 4 Calculations ---")
print(f"n = {n}, k = {k}")
print(f"Degrees of freedom (total): n - 1 = {n - 1}")
print(f"Degrees of freedom (error): n - k - 1 = {n - k - 1}")
print(f"R²_adj = 1 - (1 - {r_squared:.6f}) * ({n - 1} / {n - k - 1})")
print(f"R²_adj = {r_squared_adj:.6f}")
print(f"R²_adj as percentage = {r_squared_adj_pct:.2f}% (≈ {round(r_squared_adj_pct, 1)}%)")
```

**Why is Adjusted $R^2$ slightly different from ordinary $R^2$?**
- **Ordinary $R^2$** measures sample variance explained and naturally tends to overestimate the explanatory power in small samples ($n = 6$). It never decreases when additional predictors are added, even if they add no real explanatory value.
- **Adjusted $R^2$** introduces a penalty for the number of explanatory variables ($k$) relative to the sample size by weighting with degrees of freedom $\left(\frac{n-1}{n-k-1}\right)$. This yields an unbiased estimate of the population variance explained.
- Here, because $n$ is small ($n = 6$), the adjustment $\frac{5}{4} = 1.25$ slightly reduces the value from $91.43\%$ to $89.29\%$.


### Question 5: Calculate Predicted Values and Residuals

**Formula:**
$$\hat{y} = a + bx$$
$$\text{Residual } e = y - \hat{y}$$
Check that: $\sum e \approx 0$


```python
# Question 5: Predicted Values ŷ, Residuals e, and Summary Table

y_hat_list = []
residual_list = []

header = f"{'Month':<6} {'x (£000)':<10} {'Actual y':<10} {'Predicted ŷ':<14} {'Residual e = y - ŷ':<18}"
print(header)
print("-" * len(header))

for i in range(n):
    xi = x[i]
    yi = y[i]
    y_hat_i = a + b * xi
    e_i = yi - y_hat_i
    
    y_hat_list.append(y_hat_i)
    residual_list.append(e_i)
    
    print(f"{i+1:<6} {xi:<10} {yi:<10.1f} {y_hat_i:<14.4f} {e_i:<18.4f}")

sum_x = sum(x)
sum_y = sum(y)
sum_y_hat = sum(y_hat_list)
sum_e = sum(residual_list)

print("-" * len(header))
print(f"{'Σ':<6} {sum_x:<10} {sum_y:<10.1f} {sum_y_hat:<14.4f} {sum_e:<18.10f}")
print("-" * len(header))

print(f"\nVerification check: Σe = {sum_e:.10f} ≈ 0 (Holds exactly true!)")
```

### Question 11: Business Inference
**Question:** A manager says: “If we increase advertising, sales will definitely increase.” Do you agree with this statement? Explain your reasoning using correlation, regression, and the meaning of causation.

**Answer:**
**I do not agree with the manager’s definitive statement.**

**Reasoning:**
1. **Correlation and Regression Do Not Prove Causation:** High correlation ($r = 0.956$) and a fitted regression line establish a strong statistical association, but they do not prove that advertising directly causes the increase in sales. A lurking third variable—such as seasonal demand or macroeconomic conditions—could be driving both advertising budget expansions and higher sales simultaneously.

3. **Diminishing Returns & Market Saturation:** In business, increasing ad spend beyond a certain point typically experiences diminishing marginal returns. Sales will not continue to grow indefinitely at £1,829 per £1,000 spent.


### Question 12: Prediction ($x = 7$)
**Question:** Using your regression equation, estimate sales when advertising expenditure is $x = 7$. Is this prediction interpolation or extrapolation? Explain why this distinction matters.


```python
# Question 12: Prediction for x = 7
x_new = 7
y_pred_new = a + b * x_new

print("--- Question 12 Calculations ---")
print(f"Regression Equation: ŷ = {a:.4f} + {b:.4f} * x")
print(f"At x = {x_new}: ŷ = {a:.4f} + {b:.4f} * ({x_new}) = {y_pred_new:.4f} thousand (£{y_pred_new*1000:,.2f})")
print(f"Using rounded parameters (ŷ = 0.6 + 1.829x): ŷ = {0.6 + 1.829 * 7:.3f} thousand")
```

**Answer:**
- **Estimated sales for $x = 7$:** $\hat{y} = 0.600 + 1.8286(7) = 13.400$ (£000), which equals **£13,400**.
- **Interpolation vs. Extrapolation:** This prediction is **extrapolation**.
- **Why this distinction matters:**
  - The observed advertising spend in the dataset ranges strictly from $x = 1$ to $x = 6$. The value $x = 7$ lies outside the observed domain.
  - *Interpolation* predicts within the known range where model linearity has been empirically observed.
  - *Extrapolation* assumes the exact same linear pattern holds true beyond observed limits. This is risky in business because market saturation, competitor counter-reactions, or supply bottlenecks can alter or flatten the relationship at higher expenditure levels.

### Question 14: Logical Understanding
**Question:** An analyst says: “Because $R^2 = 91.4\%$, the model is $91.4\%$ accurate.” Is this a correct interpretation of $R^2$? Explain what is wrong with the statement.

**Answer:**
**No, this is an incorrect interpretation of $R^2$.**

**Explanation:**
1. **$R^2$ measures explained variance, not accuracy:** $R^2$ is the coefficient of determination, which represents the proportion of total variation in sales that is explained by the regression line relative to a baseline naive model (the mean $\bar{y}$).
2. **Accuracy has a different statistical definition:** Model accuracy typically refers to how close individual predictions are to actual values (e.g., Mean Absolute Percentage Error (MAPE), classification accuracy, or precision).
3. **High $R^2$ does not mean negligible errors:** A regression model can possess a very high $R^2$ (e.g., $91.4\%$) while still making substantial dollar errors on individual forecasts (e.g., Month 3 was off by £2,086 on an actual value of £4,000—a $52\%$ relative error!).
